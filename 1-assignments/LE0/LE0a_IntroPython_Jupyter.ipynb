## Python as a calculator

Python can evaluate ordinary mathematical expressions.

```python
5218 + 4683
```

pandas and NumPy also support **vectorized** calculations. Adding two Series performs the addition for every row at once.

```python
arbuthnot["boys"] + arbuthnot["girls"]
```

We can therefore plot total baptisms by year without writing a loop.

```python
total_births = arbuthnot["boys"] + arbuthnot["girls"]

plt.plot(arbuthnot["year"], total_births)
plt.xlabel("Year")
plt.ylabel("Total baptisms")
plt.title("Total baptisms in London")
plt.show()
```

The ratio of boys to girls in 1629 is:

```python
5218 / 4683
```

The same calculation for every year is:

```python
arbuthnot["boys"] / arbuthnot["girls"]
```

The **proportion** of newborns that are boys in 1629 is:

```python
5218 / (5218 + 4683)
```

and for every year:

```python
boy_proportion = arbuthnot["boys"] / (arbuthnot["boys"] + arbuthnot["girls"])
boy_proportion
```

As with a calculator, pay attention to the order of operations and use parentheses when needed.
