## Matrices

NumPy's `reshape()` creates matrices. One important difference is that NumPy normally fills arrays in **row-major** order, whereas R's `matrix()` fills by columns by default.

To reproduce R's default column-filling behavior:

````python
x = np.array([1, 2, 3, 4]).reshape((2, 2), order="F")
x
````

To fill by rows:

````python
np.array([1, 2, 3, 4]).reshape((2, 2))
````

## Random normal variables and correlation

`np.random.default_rng()` creates a modern NumPy random-number generator.

````python
rng = np.random.default_rng()

x = rng.normal(size=50)
y = x + rng.normal(loc=50, scale=0.1, size=50)

np.corrcoef(x, y)[0, 1]
````

For reproducibility, initialize the random-number generator with a fixed integer seed.

**Note:** the same seed does not generate the same numerical sequence in R and NumPy because they use different random-number generation implementations.

````python
rng = np.random.default_rng(1303)
rng.normal(size=50)
````

## Mean, variance, and standard deviation

R's `var()` and `sd()` use the sample definition with denominator $n-1$. In NumPy, specify `ddof=1` to match that convention.

````python
rng = np.random.default_rng(3)
y = rng.normal(size=100)

mean_y = np.mean(y)
var_y = np.var(y, ddof=1)
sd_y = np.std(y, ddof=1)

mean_y, var_y, np.sqrt(var_y), sd_y
````

## Contour plots

Create a two-dimensional grid with `np.meshgrid()`, compute $z=f(x,y)$ on the grid, and pass it to `plt.contour()`.

````python
x = np.linspace(-np.pi, np.pi, 50)
y = x.copy()

X, Y = np.meshgrid(x, y, indexing="ij")
F = np.cos(Y) / (1 + X**2)

plt.contour(X, Y, F)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
````

````python
plt.contour(X, Y, F, levels=45)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
````

````python
Fa = (F - F.T) / 2

plt.contour(X, Y, Fa, levels=15)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
````

## Heat maps and 3D surfaces

Matplotlib can display the same matrix as a heat map or a three-dimensional surface.

````python
plt.imshow(
    Fa,
    extent=[y.min(), y.max(), x.min(), x.max()],
    origin="lower",
    aspect="auto",
)
plt.xlabel("y")
plt.ylabel("x")
plt.title("Heat map")
plt.show()
````

````python
fig = plt.figure()
ax = fig.add_subplot(projection="3d")
ax.plot_surface(X, Y, Fa)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x, y)")
plt.show()
plt.close(fig)  # Prevent the 3-D axes from affecting later 2-D plots.
````

# 2.3.3 Indexing data

Python uses **zero-based indexing**. The first element has index `0`, the second has index `1`, and so forth.

First create the same 4×4 matrix as the R example, filling by columns:

````python
A = np.arange(1, 17).reshape((4, 4), order="F")
A
````

R's `A[2,3]` means second row, third column. The equivalent NumPy expression is:

````python
A[1, 2]
````

Select multiple rows and columns with `np.ix_()`:

````python
A[np.ix_([0, 2], [1, 3])]
````

Slices use an **exclusive** upper bound. Thus rows 1–3 and columns 2–4 in R correspond to:

````python
A[0:3, 1:4]
````

````python
A[0:2, :]
````

````python
A[:, 0:2]
````

A single row:

````python
A[0, :]
````

To omit rows or columns, `np.delete()` is clearer than R's negative-index notation.

````python
np.delete(A, [0, 2], axis=0)
````

````python
B = np.delete(A, [0, 2], axis=0)
B = np.delete(B, [0, 2, 3], axis=1)
B
````

The shape of the matrix is:

````python
A.shape
````

# 2.3.4 Loading data

pandas `read_csv()` is the standard way to import comma-separated data.

For this course, the data files are stored in `./data/`. We will use the local
`Auto.csv` file associated with *An Introduction to Statistical Learning*.
The file uses `?` for missing horsepower values, so the loader converts those
entries to pandas missing values.

````python
DATA_DIR = Path("./data")
AUTO_PATH = DATA_DIR / "Auto.csv"

if not AUTO_PATH.exists():
    raise FileNotFoundError(
        f"Could not find {AUTO_PATH.resolve()}. "
        "Place Auto.csv in a data folder in the notebook's working directory."
    )

Auto = pd.read_csv(
    AUTO_PATH,
    na_values=["?", "NA", "N/A", ""],
    skipinitialspace=True,
)

# Remove an accidental saved-index column, if the CSV was exported with one.
Auto.columns = Auto.columns.str.strip()
Auto = Auto.loc[:, ~Auto.columns.str.match(r"^Unnamed")].copy()

required_columns = [
    "mpg", "cylinders", "displacement", "horsepower", "weight",
    "acceleration", "year", "origin", "name",
]
missing_columns = [column for column in required_columns if column not in Auto.columns]
if missing_columns:
    raise ValueError(
        "Auto.csv is missing required columns: " + ", ".join(missing_columns)
    )

Auto.head()
````

The header row becomes the DataFrame's column names. It is good practice to
inspect the imported data types before beginning an analysis.

````python
Auto.dtypes
````

Inspect the number of observations and variables:

````python
Auto.shape
````

Missing values are represented by `NaN` in pandas. Count them by variable:

````python
Auto.isna().sum()
````

The original R lab removed rows containing missing values. pandas `.dropna()` does the same.

````python
Auto = Auto.dropna().copy()
Auto.shape
````

Inspect the variable names:

````python
Auto.columns
````
