**a.** A powder dispenser is set to fill packages with 250 g. An engineer wants to know whether the **mean** fill mass has dropped below 250 g.

**b.** A ventilation fan is designed to produce a mean sound level of 40 dBA. An engineer wants evidence that the mean sound level is **above** 40 dBA.

**c.** A coating process targets a mean film thickness of 12.0 $\mu$ m. Either a higher or a lower mean thickness would be a concern.

Then answer:

1. Why should the direction of $H_A$ be selected **before** looking at the sample results?
2. For scenario **b**, suppose the upper-tail $P$-value is 0.02. Interpret this value in context. Does it mean there is a 2% probability that $H_0$ is true?

Keep your response to the three hypothesis pairs and about 4–5 explanatory sentences.

    * $H_A$: $\mu > 40 [dBA]$
    * Upper tail

    * $H_A$: $\mu \neq 12.0 [\mu m]$
    * Two tails

## LE9-2. Is the adhesive taking too long to cure? (4 pts)

The file `./data/adhesive_curing_times.csv` contains the curing times (minutes) for **12 independently produced adhesive batches**. A manufacturer wants the **population mean curing time** to be **no more than 40 minutes**. Assume the batches are randomly sampled from a stable production process.

Test the boundary null model $H_0:\mu=40$ against the **upper-tail** alternative $H_A:\mu>40$. Choose $\alpha=0.05$ before examining the result.

Assume curing times are independent and come from an approximately normal population. The population SD $\sigma$ is **unknown**, so use the sample SD and Student's $t$ reference distribution.

*The datasets provided for LE09 are simulated classroom examples, not measurements from a published study.*


````python
import numpy as np
import pandas as pd
from scipy import stats

pd.set_option("display.precision", 4)

curing = pd.read_csv("./data/adhesive_curing_times.csv")
cure_times = curing["curing_time_minutes"].to_numpy()
mu0_cure = 40.0  # minutes

# Preview the source data.
display(curing.head())
````

### A. Calculate the test statistic and $P$-value (2.5 pts)

Use Python to calculate:

1. the sample size $n$, sample mean $\bar{x}$, and **sample** SD $s$ (use `ddof=1`);
2. the estimated standard error $\widehat{SE}(\bar{X})=s/\sqrt{n}$;
3. the test statistic and degrees of freedom,
   $$t=\frac{\bar{x}-40}{s/\sqrt{n}},\qquad df=n-1;$$
4. the **upper-tail** $P$-value using `stats.t.sf(t_stat, df=...)`.

Display these quantities in a labeled `pd.Series` or `pd.DataFrame`. Use the **signed** test statistic for the upper-tail probability.

**Answer:**

````python
# Your code here
# Calculate n, sample mean, sample SD, estimated SE, t, df, and P.
n = len(cure_times)
mean_cure = np.mean(cure_times)
samp_std_cure = np.std(cure_times, ddof = 1)
se_cure = samp_std_cure/np.sqrt(n)
t_stat = (mean_cure - mu0_cure)/se_cure

# Use stats.t.sf(...) for the upper-tail P-value.
p_val_cure = stats.t.sf(t_stat, df = n-1) #upper tail - survival function

# Display a labeled summary of your results.
summary_table_2A = pd.DataFrame({
    "n": n,
    "mean": mean_cure,
    "s": samp_std_cure,
    "SE": se_cure,
    "t": t_stat,
    "df": n-1,
    "p-value": p_val_cure,
},index = ["Value"])

summary_table_2A
````

### B. Make and interpret the decision (1.5 pts)

Using your calculated $P$-value:

1. Use Python to print whether you **reject** or **fail to reject** $H_0$ at the prespecified $\alpha=0.05$ and, for comparison, at the stricter $\alpha=0.01$.
2. In 1–2 sentences, explain what the result suggests about the **population mean**, and why it does **not** establish that every individual batch takes more than 40 minutes to cure.

**Answer:**

````python
# Compare your P-value with 0.05 and 0.01.
# Print the decision at each significance level.

# Your code here

alpha_init = 0.05
alpha_strict = 0.01

for alpha in alpha_init,alpha_strict:
    if p_val_cure <= alpha:
        print("Reject H0 at alpha =", alpha)
    else:
        print("Fail to reject H0 at alpha =", alpha)
````

## LE9-3. Printed trace width: statistical versus practical significance (3 pts)

The file `./data/printed_trace_widths.csv` contains trace widths (in $\mu$m), one measured trace from each of **10 independently selected printed-circuit builds**.

A printed-circuit process targets a **population mean trace width of 120 $\mu$m**. Widths that are too large **or** too small could be problematic. Assume the measurements are independent and come from an approximately normal population, with unknown $\sigma$.

The engineering tolerance for the **mean** width is **115–125 $\mu$m**. Distinguish a statistically detectable shift from a shift large enough to be of practical concern.


````python
traces = pd.read_csv("./data/printed_trace_widths.csv")
trace_width = traces["trace_width_um"].to_numpy()
mu0_width = 120.0  # micrometers

display(traces.head())
````

### A. Perform a two-sided test and construct a confidence interval (2 pts)

State $H_0$ and $H_A$ for this question. Then use `stats.ttest_1samp()` with `alternative="two-sided"` to calculate the **test statistic** and **$P$-value**.

For the same data, calculate the estimated SE using `stats.sem()` and the **95% confidence interval for the population mean** using

```python
stats.t.interval(confidence=0.95, df=n-1, loc=sample_mean, scale=estimated_se)
```

Print the statistic, $P$-value, and both interval bounds. Compare the $P$-value to $\alpha=0.05$.

**Answer:**
* $H_0: \mu = 120 [\mu m]$ 
* $H_A: \mu \neq 120 [\mu m]$

````python
# State the two hypotheses in the written-answer cell above.

# Your code here

# Run a two-sided one-sample t-test for mu0_width.
two_samp = stats.ttest_1samp(trace_width, mu0_width, alternative="two-sided")

# Calculate the estimated SE and 95% t confidence interval.
n_traces = len(trace_width)
sample_mean = np.mean(trace_width)
estimated_se = stats.sem(trace_width)
ci_traces = stats.t.interval(confidence=0.95, df = n_traces-1, loc = sample_mean, scale = estimated_se)

# Print the test statistic, P-value, CI bounds, and decision at alpha=0.05.
alpha = 0.05
print("t =", two_samp.statistic,
      "\np-value = ",two_samp.pvalue,
      "\nCI =",ci_traces)

if two_samp.pvalue <= alpha:
    print("Reject H0 at alpha =",alpha)
else:
    print("Fail to reject H0 at alpha =",alpha)
````
