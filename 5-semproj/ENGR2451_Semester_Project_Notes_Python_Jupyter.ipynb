## Choosing Datasets

**Care should be taken when choosing datasets.**

- Are there enough rows and columns?
  - A dataset that is too small may provide very little to learn.
  - A dataset that is too large may need to be subset to create a manageable scope.
- Are you interested in the topic?
- Is the dataset *too clean*?
  - You should learn to clean and handle imperfect data.
- You may need to add data from other sources.

## Databases for Open Datasets

Possible places to look for data include:

- Kaggle
- Google Dataset Search
- Open Data Stack Exchange
- Sports-Reference and other domain-specific public sources

When choosing a dataset, document its source, licensing/usage conditions, and provenance.

## Applying Visualizations

Data science is a tool to accomplish a task:

- problem solving,
- finding insights,
- communicating those insights.

Data visualizations are often the easiest way to begin understanding data:

- find problems,
- identify patterns,
- suggest potential solutions.

## Applying Visualizations: Communication

- Data visualizations are necessary for good communication.
- **No matter how strong your coding is, it does not matter unless you can communicate the result.**
- This class will practice presentations to improve communication.
- Jupyter notebooks are useful for communicating analysis from early exploration through the final report.
- Useful models generally require careful visualization and EDA first.
- Focus heavily on visualization in the first project reports.
